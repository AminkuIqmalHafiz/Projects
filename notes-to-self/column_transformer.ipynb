### **Scikit-learn column transformer**

#### **2 October 2026**
ColumnTransformer() is a Scikit-learn function that help operate multiple dataframe manipulation. It involves (as far as I know):

    - Encoding data
        - OneHotEncoder() for identity number labelling so the model do not confuse with numerical value hierarchy
        - LabelEncoder() for dependent variable, usually yes or no -> 1 or 0

    - Standardising data
        - For other data that does not require encoding, standardisation is needed to avoid model's hallucination
        - using StandardScaler() function by scikit-learn library
        
    - Imputing data
        - Missing data can be replace with any basic statistical value. e.g; mean,median
        - Using SimpleImputer() function by scikit-learn library
        - However, imputing this way is crude, I might look for better alternative (creating a small model to predict missing value)

I saw its practical use during data preprocessing, as it is able to improve code visual ,readability and structure (feels like a multi-tasking function). Here is the code snippet:




```python
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder

# Reading dataframe
vector = 'facies_vectors.csv'
df = pd.read_csv(vector)

# Target column
y = df['PE']
X = df.drop(columns=['PE'])

# Keep only rows where the target is not missing
valid_rows = y.notna()
X_train = X[valid_rows]
y_train = y[valid_rows]

# Extract test rows for specific wells
well_names = ['ALEXANDER D', 'KIMZEY A', 'Recruit F9']
mask = X['Well Name'].str.strip().str.lower().isin(
    [name.strip().lower() for name in well_names]
)
X_test = X[mask].copy()
y_test = y[mask].copy()

# Keep the well name with the target in the same table
filtered_df = df[mask][['Well Name', 'PE']].copy()

# 1. Group columns correctly, treating integer as categorical
cat_cols = ['Formation', 'Well Name', 'Facies', 'NM_M']
num_cols = ['Depth', 'GR', 'ILD_log10', 'DeltaPHI', 'PHIND', 'RELPOS']

#ColumnTransformer() in action
ct = ColumnTransformer(
    transformers=[
        ('one_hot', OneHotEncoder(sparse_output=False, drop='first'), cat_cols),
        ('scale', StandardScaler(), num_cols)
    ],
    remainder='drop' 
)

#Immediately fit into x_train
X_train_transformed = ct.fit_transform(X_train)
X_test_transformed = ct.transform(X_test)
```

The dataset here is from old ML contest back in 2016. Here, I attempted to predict PE (Photoelectric Effect) in the well logs dataset. Without the use of ColumnTransformer, It might be hard to read the code and make it way longer.
But thanks to the function, I am able to properly seperate categorical columns and numerical columns and properly allocate it to encoding and standardising.
