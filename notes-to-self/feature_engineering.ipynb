### **Feature Engineering**

#### **1 October 26**
Its a process where we use our raw data into fresh input as a buffer for the estimator's context. It is consist of giving the model the answer to the questions:

    - What is the context above the current cell?
    - What is the context below the current cell?
    - What is the change or the interval between cell?
    (which is discrete derivative between points)

Through Project 2, I discovered the fundamental needs of feature engineering into traditional Machine Learning algorithm (especially tree-based algo) like XGBOOST and Random Forest.

```python
logs = ['CALI', 'RMED', 'RDEP', 'RHOB', 'GR', 'NPHI', 'PEF', 'DTC', 'DRHO']
g = df.groupby("WELL")
depth_step = g["DEPTH_MD"].diff().replace(0, np.nan)
for c in logs :
    df[c+"_above"] = g[c].shift(1) # - the model knows what context above observed data
    df[c+"_below"] = g[c].shift(-1) # - the model knows what context below observed data
    df[c+"_slope"] = g[c].diff()/depth_step # -the model knows how fast the change going(slope, a discrete derivative)

#defining parameter and target variable
choosen_parameter = logs + [c + suffix for c in logs for suffix in ("_above", "_below", "_slope")]
target_variable = ['FORCE_2020_LITHOFACIES_LITHOLOGY']
```
