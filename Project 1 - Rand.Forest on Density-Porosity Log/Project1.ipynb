# **Using Random Forest Regressor to Estimate Density-Porosity Well Log Readings**

- This project focuses on application of machine learning onto the petrophysics,the very parameter that petroleum engineering mingled with.
- This project aims to use data-driven decision on real-world data, where potential developed machine learning algorithm proved to increase efficiency and reduce cost of the operation. 

# **1. Dataset and the source**

- Data is obtained from kaggle; the link : https://www.kaggle.com/datasets/prateekvyas/well-log-data/data
- The description about the data :
    The dataset is collected by the Kansas Geological Survey.
    It provides around 8000 data containing essential well log readings such as Resistivity, Spontaneous Potential, Gamma ray, Bulk density, Neutron and density porosity data.
- The intention behind the data is to actually build a deep learning model before apply it to a bigger datasets


# **2. The Problem**

## So, what do we actually predicting?
- We want to predict the Density-Porosity well log reading (as the title suggests) using other existing well log datasets.



## Why we want predict it in the first place?
- It is very interesting on how this particular reading was being taken. Unlike Gamma Ray and Resistivity, Density-Porosity need to utilise the use of highly radioactive material (e.g. Cesium-137). 
- If this valuable tools somehow get stuck in the well, it will cause a massive, million-dollars headache and lost of productivity time.
- So if we utilise existing data and minimise the use of this radioactive tool through machine learning, we might succeed in increasing revenue,reducing utility risk and productivity risk.
- Other than that, we also are able to prove that data-driven decision could be really useful when it comes to finding an edge in the petrophysics application. 


## Why we use Random Forest as our choice of model?
- Random Forest are actually superior in terms of hedging overfitting risk and need way less hypertuning, while proved to show high performance in evaluation metrics.
- I initally thought of using ANN regressor, but the amount of dataset given might need larger and more time is given for the hypertuning.
- This choice of model also well-regarded by the petroleum engineering society as industry-standard of machine learning in oil and gas sector.
