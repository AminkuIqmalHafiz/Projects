#### **Justification on the choosen parameter grid**

It is very important for me to address every single parameter tuned to the particular degree for cross-validate.

1) n_estimators will dictate how much desicion tree will be created for our random forest. I use 10 to 20 to determine if small amount of tree may optimise time and performance. 100 the standard, while 200 to 300 is needed if the machine wants to stabilise its R2. Our machine will choose a number that is both lightweight on computation and high performing.

2) Using max_depth will force out machine to simplify as much data pattern as possible,it perhaps may help with optimisation, but i left None if the machine thinks there is a need to be as precise to ensure performance.

3) min_samples_split and min_samples_lead is a useful tool for identifying the minimum requirement for our root to spread out. for example, 2 means that it requires at least 2 data to split. The machine will decide the best tuning to increase overall estimator performance.

4) max_features, for me, its quite interesting. Apparently, max_features will **restrict** to how much feature the tree can handle per split. 
    - **log(9) ~ 3.17 or 3, and sqrt 9 = 3**. This means that the log2 and sqrt will limit to 3 well logs reading per split only. 
    - This is to actually **eliminate potential multicollinearity between data** (which i think the selected model could untangle.)
    - I added None as a precaution if checking all 9 well logs simutaneously would aid cross-validating the machine's performance.
    - If the machine choose None, then all the well log data are assumed to be relevant for estimation and not redundant.

#### **Cross-validation grid**
- We have **405 possible grid scenarios**

**My justification on putting 5 :** 

- If I meant to choose 100 - 200, it would take couple of thousands scenario to find a suitable one (would eat up time and computational power). **40500 to 81000 adjusted possible grid scenarios**. 

- On the other hand, choosing 1-2 only would mean lack in depth for our machine to think of any better possible scenario. The estimator would be to simple and shallow to even predict our dataset. **405 to 810 adjusted possible grid scenarios**.

- I choose 5 as both computational power-friendly and adequately in-depth for the estimator.
**2025 adjusted grid scenarios**.
